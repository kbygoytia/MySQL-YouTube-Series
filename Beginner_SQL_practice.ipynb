## Import SQL and make database
- Turn your csv into a database and/or make a database from scratch
### Below we are converting CSV to database

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("layoffs.db")

pd.read_csv("layoffs.csv").to_sql("layoffs", conn, if_exists="replace", index=False)

conn.commit()
conn.close()
print("Loaded! Tables: layoffs")
```

```python
%%sql 
SELECT 
    * 
FROM 
    employee_demographics;
```

```python
%%sql 
SELECT
    first_name,
    last_name, 
    birth_date,
    age,
    (age + 10)*10
FROM 
    employee_demographics;
```

```python
%%sql 
SELECT
    DISTINCT first_name,
    gender
FROM 
    employee_demographics;
```

```python
%%sql 
SELECT 
    *
FROM 
    employee_salary
WHERE 
    # first_name = "Leslie";
    salary <= 50000;
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    gender != "Female";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    birth_date > "1985-01-01"
AND 
    gender = "Male";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    birth_date > "1985-01-01"
OR 
    gender = "Male";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    birth_date > "1985-01-01"
OR NOT 
    gender = "Male";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    (first_name = "Leslie"
AND 
    age = 44)
OR age > 55;
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    first_name LIKE "Jer%";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    first_name LIKE "%a%";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    first_name LIKE "a%";
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
WHERE
    birth_date LIKE "1989%";
```

```python
%%sql
SELECT 
    gender,
    AVG(age)
FROM
    employee_demographics
GROUP BY 
    gender;
```

```python
%%sql
SELECT 
    gender,
    AVG(age),
    MAX(age),
    MIN(age),
    COUNT(age)
FROM
    employee_demographics
GROUP BY 
    gender;
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
ORDER BY first_name;
```

```python
%%sql
SELECT 
    *
FROM
    employee_demographics
ORDER BY first_name DESC;
```

```python
%%sql
SELECT 
    *
FROM 
    employee_demographics
ORDER BY 
    gender, age DESC;
```
